```python
class Animal(object):
    # constructor
    def __init__(self, age):
        self._age = age 
        self._name = None    

    # getter attribute age
    @property
    def age(self):
        return self._age   #protected attribute
    
    # setter attribute age
    @age.setter
    def age(self, newage):
        if newage < 0 :
            raise ValueError('age cannot be negative')        
        self._age = newage

    # getter attribute name
    @property   
    def name(self):
        return self._name  #protected attribute
    
    # setter attribute name
    @name.setter
    def name(self, newname):
        self._name = newname

    def __str__(self):
        return f'Animal : {self.name}, {self.age}'
```

```python
animal1 = Animal(3)
print(animal1)
animal1.name = 'Magu'
animal1.age += 1
print(animal1)
print(animal1.age) 
```

```python
class Cat(Animal):
    def speak(self):
        print('Meau Meau')

    def __str__(self):
        return f'Cat is a {super().__str__()}'
```

```python
cat1 = Cat(1)
print(cat1)
cat1.name = 'Jaru'
print(cat1)
cat1.speak()
cat1.age = 5
```

```python
class Person(Animal):
    # Constructor
    def __init__(self, age, name):
        super().__init__(age)       # Animal.__init__(self, age)                  call constructor of parent
        self.name = name        # call @name.setter of name in Animal
        self._friends = []   # initialize protected attribute

    # getter _friends attribute
    @property
    def friends(self):
        return self._friends.copy()
    
    # setter _friends attribute
    @friends.setter
    def friends(self, friend):
        if friend not in self._friends:
            self._friends.append(friend)

    def __str__(self):
        return f'Person is a {super().__str__()}'
    
    def age_diff(self, other):
        diff = self.age - other.age
        return f'{diff} years difference'  
```

```python
p1 = Person(54, 'Wei')
print(p1)
p1.friends = 'Gu'
p1.friends = 'Ming'
p1.friends = 'Ming'
print(p1.friends)
```

```python
def make_pets(d):
    for person, cat in d.items():
        print(f"{person.name}:{cat.name}")
```

```python
class Student(Person):
    def __init__(self, age, name, major):
        super().__init__(age, name)     # Person.__init__(self, age, name)
        self.major = major      #call setter 
    
    @property
    def major(self):
        return self.__major     #private attribute
    
    @major.setter
    def major(self, newmajor):
        self.__major = newmajor

    def __str__(self):
        return f'Student is a {super().__str__()} , {self.major} '
    
    def speak(self):
        print(f'I am {self.name}, {self.age} years old, I am student in {self.__major}')
```

```python
s1 = Student(45, 'Hing', 'comp')
s1.friends = 'Gin'
print(s1)
s1.speak()
```
